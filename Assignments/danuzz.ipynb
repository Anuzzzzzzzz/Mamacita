```python

marks = 72

if marks >= 80:
    grade = "A"
elif marks >= 60:
    grade = "B"
elif marks >= 40:
    grade = "C"
else:
    grade = "Fail"

print(f"Marks: {marks} → Grade: {grade}")
```

```python
# Access Control Example

is_student = True
has_id = True
in_dress = False

if is_student and has_id and in_dress:
    print("Entry allowed. Can enter the classroom.")

elif is_student and in_dress or has_id:
    print("Partial entry allowed!")
    if is_student and has_id:
        print("Not allowed in classroom.")

else:
    print("Not allowed.")
```

```python
# Temperature Converter with History

history = []

def temperature_menu():
    print("\nTemperature Converter")
    print("1. Celsius to Fahrenheit")
    print("2. Fahrenheit to Celsius")
    print("3. Celsius to Kelvin")
    print("4. Kelvin to Celsius")
    print("5. Exit")

while True:
    temperature_menu()
    choice = input("Enter choice (1-5): ")

    if choice == "5":
        break

    value = float(input("Enter temperature value: "))

    if choice == "1":
        result = celsius_to_fahrenheit(value)
        unit = "°F"
    elif choice == "2":
        result = fahrenheit_to_celsius(value)
        unit = "°C"
    elif choice == "3":
        result = celsius_to_kelvin(value)
        unit = "K"
    elif choice == "4":
        result = kelvin_to_celsius(value)
        unit = "°C"
    else:
        print("Invalid choice")
        continue

    history.append(result)
    print(f"Converted Value: {result:.2f} {unit}")

    if result <= 0:
        print("⚠ Freezing Point Warning!")
```
