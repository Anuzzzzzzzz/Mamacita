```python
# *args and **kwargs
# Use *args to pass multiple positional arguments.

def show_args(*args):
    global lists
    lists =0
    for item in args:
        print(item)
        lists += int(item)
    return lists

value =show_args("111", "222")
print(value)
print(lists)
```

```python
# Use **kwargs to pass multiple keyword arguments.

name = input("Enter the Name:")
def show_kwargs(**kwargs):
    for key, value in kwargs.items():
        print(f"{key} = {value}")

show_kwargs(name=name, age=23, surname= "Paudel", level="999")
show_kwargs(name=name, age=24, surname= "Sharma", level="999")
```
