```python
### Defining Methods

# Methods are functions defined inside a class that define the object's behavior. They also take `self` as the first parameter,
# allowing them to access and modify the object's attributes.


class Dog:
    def __init__(self, name, breed, age):
        self.name = name
        self.breed = breed
        self.age = age
        print(f"A new Dog object named {self.name} was created!")


    def bark(self):
        print(f"{self.name} says: Woof! Woof!")

    def describe(self , location):
        print(f"This is {self.name}, a {self.age}-year-old {self.breed}, normally found in {location}.")

    def have_birthday(self):
        self.age += 1
        print(f"Happy Birthday, {self.name}! You are now {self.age} years old.")
```

```python
# ### Defining Superclass and Subclass
# Let's define a basic `Person` class.

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        print(f"Person object created: {self.name}")

    def introduce(self):
        print(f"Hi, my name is {self.name} and I am {self.age} years old.")

    def walk(self):
        print(f"{self.name} is walking.")

# Create a Person object
p1 = Person("Alice", 30)
p1.introduce()
p1.walk()
```

```python
# 1. Define the Parent Class
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"Hi, I am {self.name} and I am {self.age} years old.")

    def walk(self):
        print(f"{self.name} is walking.")

# 2. Define the Child Class (Inherits from Person)
class Student(Person):
    # We can add new methods specific only to Students
    def study(self):
        print(f"{self.name} is studying for the exam.")

# 3. Create an instance of the Child class
s1 = Student("Bob", 20)

# Accessing attributes and methods from the Parent class
print(f"Student's Name: {s1.name}")
s1.introduce()
s1.walk()
```

```python
# ### Adding Specific Attributes/Methods to Subclass


class Student(Person):
    def __init__(self, name, age, student_id, major):
       
        super().__init__(name, age)
        self.student_id = student_id
        self.major = major

    def study(self):
        print(f"{self.name} (ID: {self.student_id}) is studying {self.major}.")

s2 = Student("Charlie", 21, "S12345", "Computer Science")


print(f"\nName: {s2.name}, Age: {s2.age}")
s2.walk()

# Access new attributes/methods
print(f"Student ID: {s2.student_id}, Major: {s2.major}")
s2.study()
```
