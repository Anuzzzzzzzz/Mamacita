```python
# ##### Statistical Functions

# `NumPy` provides built-in methods to compute statistics like mean, sum, and standard deviation.
```

```python

import numpy as np
data = np.array([[80, 85, 90],
                 [70, 75, 65],
                 [88, 92, 95]])

print("Mean:", np.mean(data))
print("Sum:", np.sum(data))

axis1 = np.mean(data, axis=1) # Mean across rows
print("Mean across rows (axis=1):", axis1)

axis0 =np.mean(data, axis=0) # Mean across columns
print("Mean across columns (axis=0):", axis0)
print("Standard Deviation:", np.std(data))
print("Max of data:", np.max(data) )
print("Min of data:", np.min(data) )


```

```python
student1_mean = np.mean(data[0])
print("Mean of student 1:", student1_mean)

student2_mean = np.mean(data[1])
print("Mean of student 2:", student2_mean)

student3_mean=np.mean(data[2])
print("Mean of student 3:",student3_mean)
```

```python
#mean for 2nd and 3rd student in same
mean_2_3 = np.mean(data[1:3], axis=1)
print("Mean of student 2 and 3:", mean_2_3)

```

```python
# ##### Boolean Indexing & Filtering

# - Use boolean masks to filter arrays based on conditions.

# - In NumPy, you can pass a boolean array inside square brackets to filter the original array. This is called boolean indexing.
```

```python
scores = np.array([55, 72, 89, 45, 99, 66, 78])
high_scores = scores[scores > 70]
print("Scores > 70:", high_scores)

# Find even scores
even_scores = scores[scores % 2 == 0]
print("Even Scores:", even_scores)

# Find scores between 60 and 80
mid_scores = scores[(scores >= 60) & (scores <= 80)]
print("Scores between 60 and 80:", mid_scores)
```

```python
high = data[data > 70]
print("high scores 2D array:", high)
```

```python
import numpy as np
np.random.seed(42)
student_scores = np.random.randint(50, 100, size=(5, 4))

print("All Scores:\n", student_scores)
print("-" * 20)

print("Averages:", np.mean(student_scores, axis=1))
print("-" * 20)

# Logic: If a student's MAX score is > 75, keep that row.
print("Students > 75:\n", student_scores[np.max(student_scores, axis=1) > 75])
```

```python
# ## Assignment: Analyze Numeric Data and Filter Top 10% Scores

# Given a list of exam scores for 30 students, write a program to:
# - Compute the mean and standard deviation
# - Filter scores in the top 10%

import numpy as np

np.random.seed(42)

# 30 random scores between 50 and 100
scores = np.random.randint(50, 101, 30)

print("Scores:", scores)


```

```python
mean = np.mean(scores)
std_dev = np.std(scores)

print("Mean:", mean)
print("Standard Deviation:", std_dev)

```

```python
top_10 = np.sort(scores)[-3:]
print("Top 10% Scores:", top_10)
```
