```python
# #### Slicing and Reshaping Arrays

# * Slicing helps extract parts of arrays.
# * Reshaping changes the shape without changing data.
```

```python
import numpy as np

arr = np.arange(12)
print("Original:", arr)

#Slicing
print("First 5:", arr[:5])
print("Last 3:", arr[-3:])
arr.shape
```

```python
arr.ndim
```

```python
reshaped = arr.reshape(3, 4)
print("Reshaped (3x4):\n", reshaped)

print("\nLast column:", reshaped[:, -1])
print("First two rows:\n", reshaped[:-1])

reshaped[0][-1]=55
arr.ndim
```

```python
reshaped[1,1]
```

```python
reshaped[0, -1] = 55
```

```python
reshaped
```

```python
reshaped.ndim
```

```python
reshaped_arr = arr.reshape((3, 4))
print("Reshaped (3x4):\n", reshaped_arr)
reshaped_arr
reshaped_arr[2, 3]
reshaped_arr[2, 3] = 99
reshaped_arr

```

```python
reshaped_arr.shape
```

```python
reshaped_arr.ndim
```

```python
maximum = np.max(reshaped_arr)
print("Maximum value:", maximum)
```

```python
minimum = np.min(reshaped_arr)
print("Minimum value:", minimum)
```

```python
# ##### Vectorized Operations & Broadcasting

# * `Vectorized operations` allow fast element-wise operations.
# * `Broadcasting` enables operations between arrays of different shapes.
```

```python
x = np.array([1, 2, 3])
y = np.array([4, 5, 6])
print("Addition:", x + y)
print("Multiplication:", x * y)
print("Scalar Multiplication:", x * 10)

mat = np.ones((2,3), dtype=int)
print( mat)


print("-------")
vec = np.array([1, 2, 3])

add= mat + vec
mul=mat @ vec

print("Broadcasted Multiplication:\n", mat @ vec)


print("Broadcasted Addition:\n", mat + vec)
add.shape
mul.shape

```
