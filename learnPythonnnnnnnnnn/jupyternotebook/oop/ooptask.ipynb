```python
# Task 1:

# Create a class called Book
# with the following attributes: title, author,
# price. Write a method display_info() that prints all the details of the book. Create two objects
# of the class and display their information.

class Book:
    def __init__(self, title, author, price):
        self.title = title
        self.author = author
        self.price = price

    def display_info(self):
        print(f"Title: {self.title}")
        print(f"Author: {self.author}")
        print(f"Price: ${self.price:.2f}")
# Creating two objects of the Book class
book1 = Book("The wind", "Anuz", 999)
book2 = Book("How to be happy", "Happy Man", 799)
# Displaying their information
book1.display_info()
print()  # Just for better readability
book2.display_info()

```

```python
import math

class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * (self.radius ** 2)

    def circumference(self):
        return 2 * math.pi * self.radius

# Creating a circle object
circle = Circle(5)

# Printing its area and circumference
print(f"Area of the circle: {circle.area():.2f}")
print(f"Circumference of the circle: {circle.circumference():.2f}")
```

```python
    # Task 3:

    # Design a class Student with attributes name, roll_number, and marks
    # (list of 5 subject marks). Write a method to calculate the average marks
    # and grade of the student (grade based on average: A ≥ 90, B ≥ 75, C ≥ 50, F
    # < 50). Test your class with one student object. 


class Student:    
    def __init__(self, name, roll_number, marks):
        self.name = name
        self.roll_number = roll_number
        self.marks = marks

    # This allows s1.average() to work
    def average(self):
        return sum(self.marks) / len(self.marks)

    # Optional: grade method
    def grade(self):
        avg = self.average()
        if avg >= 90:
            return 'A'
        elif avg >= 75:
            return 'B'
        elif avg >= 50:
            return 'C'
        else:
            return 'F'
        
    def summary(self):
        print(f"Student Name: {self.name}")
        print(f"Roll Number: {self.roll_number}")
        print(f"Marks: {self.marks}")
        print(f"Average: {self.average():.2f}")
        print(f"Grade: {self.grade()}")
        
s1 = Student("Panuz", 7, [99, 92, 99, 99, 90])

```

```python
s1.summary()
```
