```python
class Vehicle:
    def __init__(self, make, model):
        self.make = make
        self.model = model

class Car(Vehicle):
    def __init__(self, make, model,doors):
        super().__init__(make, model)
        self.doors = doors

# Create a Car object
car = Car("BMW", "M5", 4)
```
