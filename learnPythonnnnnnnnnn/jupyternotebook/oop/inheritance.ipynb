```python

# Create a base class Vehicle with attributes make and model
# and a method display_info(). Inherit from this class to create a Car class with an additional attribute num_doors. 
# Demonstrate creating a Car object and calling display_info().

class Vehicle:
    def __init__(self, make, model):
        self.make = make
        self.model = model

    def display_info(self):
        print(f"Make: {self.make}, Model: {self.model}")

class Car(Vehicle):
    def __init__(self, make, model, num_doors):
        super().__init__(make, model)   # Call parent constructor
        self.num_doors = num_doors

    def display_info(self):
        super().display_info()          # Call parent method
        print(f"Number of Doors: {self.num_doors}")

c = Car("Bmw", "M5", 4)
c.display_info()
```
