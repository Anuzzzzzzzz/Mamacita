```python
df=df.drop_duplicates()
```

```python
df['company_name'] = (
    df['company_name']
    .astype(str)
    .str.lower()
    .str.strip()
)
```

```python
top_companies = df['company_name'].value_counts().head(10)
top_companies
```

```python
df = df.drop(columns=[
    'closed_time',
    'skills_desc',
    'fips',
    'zip_code'
])
```

```python
df['formatted_experience_level'] = (
    df['formatted_experience_level']
    .fillna('unknown')
    .str.lower()
    .str.strip()
)
```

```python
df
```

```python
df['salary_raw'] = df['med_salary'].combine_first(df['max_salary'])
```

```python
salary_df = df[df['salary_raw'].notna()].copy()
```

```python
df
```

```python
salary_df['pay_period'] = salary_df['pay_period'].fillna('unknown').str.lower().str.strip()
```

```python
salary_df = salary_df[
    salary_df['pay_period'].isin(['hourly', 'weekly', 'monthly', 'yearly'])
].copy()
```

```python
def salary_to_yearly(row):
    salary = row['salary_raw']
    period = row['pay_period']

    if period == 'hourly':
        return salary * 40 * 52
    elif period == 'weekly':
        return salary * 52
    elif period == 'monthly':
        return salary * 12
    elif period == 'yearly':
        return salary
```

```python
salary_df['salary_yearly'] = salary_df.apply(salary_to_yearly, axis=1)
```

```python
salary_df = salary_df[
    (salary_df['salary_yearly'] >= 10_000) &
    (salary_df['salary_yearly'] <= 500_000)
]
```

```python
plt.figure(figsize=(12,6))
sns.histplot(
    salary_df['salary_yearly'],
    bins=40,
    kde=True
)
plt.xlabel('Yearly Salary')
plt.ylabel('Number of Jobs')
plt.title('Yearly Salary Distribution')
plt.show()
```

```python
plt.figure(figsize=(12,5))
sns.boxplot(
    x=salary_df['salary_yearly']
)
plt.xlabel('Yearly Salary')
plt.title('Yearly Salary Boxplot')
plt.show()
```

```python
plt.figure(figsize=(14,6))
sns.boxplot(
    data=salary_df,
    x='formatted_experience_level',
    y='salary_yearly'

)
plt.xticks(rotation=30)
plt.xlabel('Experience Level')
plt.ylabel('Yearly Salary')
plt.title('Salary vs Experience Level')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.boxplot(
    data=salary_df,
    x='pay_period',
    y='salary_yearly'
)
plt.xlabel('Pay Period')
plt.ylabel('Yearly Salary')
plt.title('Converted Salary by Pay Period')
plt.show()
```
