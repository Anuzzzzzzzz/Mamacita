```python
df=df.drop_duplicates()
```

```python
df['company_name'] = (
    df['company_name']
    .astype(str)
    .str.lower()
    .str.strip()
)
```

```python
df = df.drop(columns=[
    'closed_time',
    'skills_desc',
    'fips',
    'zip_code'
])
```

```python
df['formatted_experience_level'] = (
    df['formatted_experience_level']
    .fillna('unknown')
    .str.lower()
    .str.strip()
)
```

```python
df
```

```python
df['salary_raw'] = df['med_salary'].combine_first(df['max_salary'])
```

```python
salary_df = df[df['salary_raw'].notna()].copy()
```

```python
df
```

```python
salary_df['pay_period'] = salary_df['pay_period'].fillna('unknown').str.lower().str.strip()
```

```python
def salary_to_yearly(row):
    salary = row['salary_raw']
    period = row['pay_period']

    if period == 'hourly':
        return salary * 40 * 52
    elif period == 'weekly':
        return salary * 52
    elif period == 'monthly':
        return salary * 12
    else:  # yearly or unknown
        return salary

salary_df['salary_yearly'] = salary_df.apply(salary_to_yearly, axis=1)
```
