Filtering:

```python
### Filtering

# Method 1: Boolean Masking
mask = (df['Price'] > 990) & (df['Category'] == 'Electronics')
print(f"High Value of Electronics Orders (Mask): {len(df[mask])}")
```

```python
mask = (df['Customer Gender'] == 'Female') & (df['Total Sales'] > 900)
print(f"Count (Mask): {len(df[mask])}")
```

```python
#method 2: Query syntax

filtering_query = df.query('Price>990 & Category == "Electronics"')
print(f"High Value of Electronics Orders (Query): {len(filtering_query)}")
display(filtering_query[['Order ID', 'Price']].head(3))
```

```python
query_df = df.query("`Customer Gender` == 'Female' and `Customer Age` < 20")
print(f"Count (Query): {len(query_df)}")

display(query_df.head(3))
```

```python
def classify_age(age):
    if age <30:
        return 'Young'
    elif age <50:
        return 'Middle-aged'
    else:
        return 'Senior'
    #apply to customer age column
df['Age Group'] = df['Customer Age'].apply(classify_age)
df[['Customer Age', 'Age Group']].head(10)

#verify
print(df['Age Group'].value_counts())
display(df[['Customer Age', 'Age Group']].head(10))
```

```python
#grouping and aggregating
grouped_df = df.groupby('Category').agg({
    'Total Sales': 'sum',
    'Order ID': 'count'
}).rename(columns={'Order ID': 'Number of Orders'})
display(grouped_df)
```

```python
# Group by Category, Sum the Total Sales
cat_performance = df.groupby('Category')['Total Sales'].sum().reset_index()

# Sort for better view
cat_performance = cat_performance.sort_values('Total Sales', ascending=False)

# Format as currency string for readability
pd.options.display.float_format = '${:,.2f}'.format

display(cat_performance)
```

```python
aged_df = df.groupby('Age Group').agg({
    'Total Sales': 'sum',
    'Order ID': 'count'
}).rename(columns={'Order ID': 'Number of Orders'})
display(aged_df)
```

```python
cat_performance = df.groupby(['Category']).agg({
    'Total Sales': ['sum', 'mean', 'max'],
    'Order ID': 'count'
})
display(cat_performance)
```

```python
# ### B. Advanced Aggregation
# **Task:** Analyze `Total Sales` by `Age Group`.
#
# We want: Min, Max, Average, and Count of sales.
age_group_performance = df.groupby('Age Group').agg({
    'Total Sales': ['min', 'max', 'mean'],
    'Order ID': 'count'
})
display(age_group_performance)
```
