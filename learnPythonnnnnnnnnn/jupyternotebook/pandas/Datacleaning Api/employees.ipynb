```python
#cleaning
df=df.dropna(subset=['name'])
df
```

```python
mean_age=df['age'].mean()
median_salary=df['salary'].median()
df=df.fillna({'age':mean_age,'salary':median_salary})
df=df.fillna({'department':'unknown'})
df
```

```python
df['age']=df['age'].astype(int)
df['salary']=df['salary'].astype(int)
```

```python
df['bonus']=0.1*df['salary'].astype(int)
df
```

```python
df['performance_score']=np.random.randint(1,6,59)
df
```

```python
df['join_date']=df['join_date'].ffill()
df
```
