"Matplotlib" is a low level data visualization lib that is build in
NumPy arrays. It consists of various plots like scatter plot,line plot,histogram etc

Scatter plot

```python
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("Pandas/tips.csv")

scatter = plt.scatter(
    data["day"],
    data["tip"],
    c=data["total_bill"],
)

plt.colorbar(scatter, label="Total Bill")
plt.title("Tips by Day")
plt.xlabel("Day")
plt.ylabel("Tip")
plt.show()
```

Line Chart

```python
import pandas as pd
import matplotlib.pyplot as plt
data = pd.read_csv("Pandas/tips.csv")

avg_tip = data.groupby('size')['tip'].mean()

plt.plot(avg_tip.index, avg_tip.values, marker='o')
plt.title("Average Tip by Group Size")
plt.xlabel("Size")
plt.ylabel("Average Tip")
plt.show()


plt.show()
```

Bar chart

```python
import pandas as pd
import matplotlib.pyplot as plt
data = pd.read_csv("Pandas/tips.csv")
plt.bar(data['day'], data['tip'])
plt.title("Tips by Day")
plt.xlabel("Day")
plt.ylabel("Tip")
plt.show()
```

Histogram

```python
import pandas as pd
import matplotlib.pyplot as plt
data = pd.read_csv("Pandas/tips.csv")
plt.hist(data['total_bill'])
plt.title("Total Bill")
plt.show()
```

Bokeh
"Bokeh" is mainly famous for its interactive charts visualization.
it renders its plots using html and js


Scatter plot

```python
# importing the modules
from bokeh.plotting import figure, output_file, show
from bokeh.palettes import magma
import pandas as pd


# instantiating the figure object
graph = figure(title = "Bokeh Scatter Graph")

# reading the database
data = pd.read_csv("pandas/tips.csv")

color = magma(256)

# plotting the graph
graph.scatter(data['total_bill'], data['tip'], color=color)

# displaying the model
show(graph)
```

Line Chart

```python
# importing the modules
from bokeh.plotting import figure, output_file, show
import pandas as pd


# instantiating the figure object
graph = figure(title = "Bokeh Bar Chart")

# reading the database
data = pd.read_csv("pandas/tips.csv")

# Count of each unique value of
# tip column
df = data['tip'].value_counts()

# plotting the graph
graph.line(df, data['tip'])

# displaying the model
show(graph)
```

BAR CHART


```python
# importing the modules
from bokeh.plotting import figure, output_file, show
import pandas as pd


# instantiating the figure object
graph = figure(title = "Bokeh Bar Chart")

# reading the database
data = pd.read_csv("pandas/tips.csv")

# plotting the graph
graph.vbar(data['total_bill'], top=data['tip'])

# displaying the model
show(graph)
```

INTERACTIVE DATA VISUALIZATION #hiding/muting

```python
# importing the modules
from bokeh.plotting import figure, output_file, show
import pandas as pd


# instantiating the figure object
graph = figure(title = "Bokeh Bar Chart")

# reading the database
data = pd.read_csv("pandas/"
                   "tips.csv")

# plotting the graph
graph.vbar(data['total_bill'], top=data['tip'],
           legend_label = "Bill VS Tips", color='green')

graph.vbar(data['tip'], top=data['size'],
           legend_label = "Tips VS Size", color='red')

graph.legend.click_policy = "hide"

# displaying the model
show(graph)
```
