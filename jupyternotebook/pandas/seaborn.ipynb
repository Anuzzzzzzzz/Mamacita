```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import io

csv_data = """ProductID,ProductName,Category,Price,UnitsSold
101,Laptop,Electronics,1200,50
102,Mouse,Electronics,25,200
103,Keyboard,Electronics,75,150
104,T-Shirt,Apparel,20,300
105,Jeans,Apparel,95,120
106,Coffee Mug,Homeware,15,500
"""

df = pd.read_csv(io.StringIO(csv_data)) # Load CSV data from a string using StringIO as a file-like object

print(df["ProductName"])
```

```python
df['Revenue'] = df['Price'] * df['UnitsSold']
```

```python
# Now let's plot the total revenue by category
plt.figure(figsize=(8, 6))

category_revenue = df.groupby('Category')['Revenue'].sum().reset_index()

sns.barplot(x='Category', y='Revenue', data=category_revenue, hue='Revenue',palette='Spectral')

plt.title('Total Revenue by Product Category')
plt.xlabel('Category')

plt.ylabel('Total Revenue ($)')
plt.show()
```

```python
sns.lineplot(x='Category', y='Revenue', data=category_revenue, marker='o')

plt.title('Total Revenue by Product Category')
plt.xlabel('Category')
plt.ylabel('Total Revenue ($)')
plt.show()
```

```python
import matplotlib.pyplot as plt

# Pie chart for Units Sold per Product
plt.figure(figsize=(8, 8))

plt.pie(df['UnitsSold'],
        labels=df['ProductName'],
        autopct='%1.1f%%',     #
        startangle=0)

plt.title('Units Sold Distribution per Product')
plt.tight_layout()  # Prevent overlap
plt.legend(loc='best')
plt.show()
```

```python
# --- Your Turn: Visualization Mini-Project ---


plt.figure(figsize=(10, 6))
sns.barplot(x='Revenue',y='ProductName', hue='Revenue', data=df.sort_values('Revenue', ascending=False),palette='viridis')
plt.title('Total Revenue per Product')
plt.xlabel('Total Revenue ($)')
plt.ylabel('Product Name')
plt.show()
```

```python
# Plot 2: How do the categories compare in terms of units sold?
category_units = df.groupby('Category')['UnitsSold'].sum().reset_index()
plt.figure(figsize=(8, 8))
plt.pie(category_units['UnitsSold'], labels=category_units['Category'], autopct='%1.1f%%', startangle=0, colors=sns.color_palette('pastel'))
plt.title('Share of Total Units Sold by Category')
plt.ylabel('')
plt.show()
```

```python
daily_gender = df.groupby(['Date', 'Gender'])['Sales'].sum().reset_index()

plt.figure(figsize=(12, 6))
sns.lineplot(data=daily_gender, x='Date', y='Sales', hue='Gender', marker='o')
plt.title('Daily Sales by Gender')
plt.xlabel('Date')
plt.ylabel('Sales ($)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
product_sales = df.groupby('Product line')['Sales'].sum().reset_index()

plt.figure(figsize=(10, 6))
sns.barplot(data=product_sales, x='Product line', y='Sales', palette='viridis')
plt.title('Total Sales by Product Line')
plt.xticks(rotation=45)
plt.ylabel('Sales ($)')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.countplot(data=df, x='Customer type', hue='Payment', palette='Set2')
plt.title('Customer Type vs Payment Method')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))
sns.countplot(data=df, x='Date', hue='Gender', palette='coolwarm')
plt.title('Transactions by Date and Gender')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
