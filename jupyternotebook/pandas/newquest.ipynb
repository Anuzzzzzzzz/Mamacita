```python
# 1.  **Create and Load the Data:** Run the provided code to create `student_performance_messy.csv` and load it into a DataFrame.
# 2.  **Inspect the Data:** Use `.info()`, `.head()`, and `.isnull().sum()` to understand the issues.
# 3.  **Clean the Data:**
#     *   The `gender` column has inconsistent values (`M`, `F`, `Male`, `Female`). Standardize them to `M` and `F`.
#     *   The `score` column has some values as percentages (e.g., "75%") and some missing. Convert all scores to a numeric format (e.g., 75.0).
#     *   Handle the missing scores by filling them with the **mean** score of their respective `gender`. (This is a bit more advanced and requires grouping!).
#     *   There is a `study_hours` column with some negative values, which is impossible. Replace any negative hours with 0.
#     *   Check for and remove any duplicate rows.
# # 4.  **Export the Result:** Save your final, cleaned DataFrame to a file named `student_performance_cleaned.csv`.
# import pandas as pd
#
# students_data_messy = """student_id,name,gender,score,study_hours
# 1,Alice,F,85,10
# 2,Bob,M,92,8
# 3,Charlie,Male,"78%",12
# 4,Diana,F,,15
# 5,Ethan,M,65,5
# 6,Fiona,Female,95,11
# 7,George,M,-5,4
# 5,Ethan,M,65,5
# 8,Hannah,F,88,9
# 9,Ian,Male,"62%",-2
# 10,Jane,F,,14
# """
#
# with open('student_performance_messy.csv', 'w') as f:
#     f.write(students_data_messy)
#
# print("Assignment dataset 'student_performance_messy.csv' created!")
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv('student_performance_messy.csv')

# Inspect columns
print(df.columns.tolist())
df.info()
df.head()
df.isnull().sum()

# Clean gender
df['gender'] = df['gender'].replace({'Male': 'M', 'Female': 'F'})

# Clean score
df['score'] = df['score'].str.replace('%', '').astype(float)

# Fill missing scores with mean by gender
df['score'] = df.groupby('gender')['score'].transform(lambda x: x.fillna(x.mean()))

# Fix negative study_hours
df['study_hours'] = df['study_hours'].apply(lambda x: x if x >= 0 else 0)

# Remove duplicates
df = df.drop_duplicates()

# Export cleaned CSV
df.to_csv('student_performance_cleaned.csv', index=False)
print("Cleaned CSV saved as 'student_performance_cleaned.csv'")

# Check result
print(df)
```
