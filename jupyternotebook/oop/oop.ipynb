```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"Hi, I'm {self.name} and I'm {self.age} years old.")

    def walk(self):
        print(f"{self.name} is walking.")


class Student(Person):
    def __init__(self, name, age):
        super().__init__(name, age)


s1 = Student("Bob", 20)

print(f"Student's Name: {s1.name}")
s1.introduce()
s1.walk()

print(f"\nType of s1: {type(s1)}")
print(f"Is s1 an instance of Student? {isinstance(s1, Student)}")
print(f"Is s1 an instance of Person? {isinstance(s1, Person)}")
```

```python
# 1. Define the Parent Class
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        print(f"Hi, I am {self.name} and I am {self.age} years old.")

    def walk(self):
        print(f"{self.name} is walking.")

# 2. Define the Child Class (Inherits from Person)
class Student(Person):
    # We can add new methods specific only to Students
    def study(self):
        print(f"{self.name} is studying for the exam.")

# 3. Create an instance of the Child class
s1 = Student("Bob", 20)

# Accessing attributes and methods from the Parent class
print(f"Student's Name: {s1.name}")
s1.introduce()
s1.walk()
```

```python
# ### Adding Specific Attributes/Methods to Subclass


class Student(Person):
    def __init__(self, name, age, student_id, major):
       
        super().__init__(name, age)
        self.student_id = student_id
        self.major = major

    def study(self):
        print(f"{self.name} (ID: {self.student_id}) is studying {self.major}.")

s2 = Student("Charlie", 21, "S12345", "Computer Science")


print(f"\nName: {s2.name}, Age: {s2.age}")
s2.walk()

# Access new attributes/methods
print(f"Student ID: {s2.student_id}, Major: {s2.major}")
s2.study()
```

```python
class Student(Person): # Inherits from Person
    def __init__(self, name, age, student_id, major):
        super().__init__(name, age) # Initialize parent attributes
        self.student_id = student_id
        self.major = major
        print(f" -> Student object initialization complete for {self.name}")

    def study(self):
        print(f"{self.name} (ID: {self.student_id}) is studying {self.major}.")

    def introduce(self):
        super().introduce()
       
        print(f"I am a student majoring in {self.major} with ID {self.student_id}.")

# Create objects
p2 = Person("David", 45)
s3 = Student("Eve", 19, "S67890", "Physics")

print("\n--- Introductions ---")
p2.introduce() # Calls Person's introduce
print("-" * 10)
s3.introduce() # Calls Student's overridden introduce
```

```python
class BankAccount:
    def __init__(self, owner, initial_balance=0):
        self.owner = owner
        self._balance = initial_balance
        self.__pin = "5678"
        print(f"Account created for {self.owner}")

    def intro(self):
        print(f"Owner Name:{self.owner}")
       
    # Public method to deposit money
    def deposit(self, amount):
        if amount > 0:
            self._balance += amount
            print(f"Deposited ${amount}. New balance: ${self._balance}")
        else:
            print("Deposit amount must be positive.")

    # Public method to withdraw money
    def withdraw(self, amount):
        if amount <= 0:
            print("Withdrawal amount must be positive.")
        elif amount > self._balance:
            print(f"Insufficient funds. Cannot withdraw ${amount}. Balance is ${self._balance}")
        else:
            self._balance -= amount
            print(f"Withdrew ${amount}. New balance: ${self._balance}")

    # Public method to get balance (controlled access)
    def get_balance(self):
        print(f"Current balance for {self.owner}: ${self._balance}")
        # We can access the "private" attribute internally
        # print(f"   (Pin : {self.__pin})")
        return self._balance

    def get_pin(self):
        if self._BankAccount__pin == self.__pin:
            print(id(self._BankAccount__pin))
            print(id(self.__pin))
           
            print(f"   (Current pin  is: {self._BankAccount__pin})")
        return self.__pin
```

```python
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
        print(f"Person object created: {self.name}")

    def introduce(self):
        print(f"Hi, my name is {self.name} and I am {self.age} years old.")

class Teacher(Person):
   
    def __init__(self, name, age, employee_id, subject):
        super().__init__(name, age)
        self.employee_id = employee_id
        self._subject = subject        
        print(f" -> Teacher object initialization complete for {self.name}")

    def introduce(self):
        super().introduce() # Call parent's introduce
        print(f"   I am a teacher with employee ID {self.employee_id} and I teach {self._subject}.")


    def teach(self):
        print(f"{self.name} is teaching {self._subject}.")
```

```python
teacher1 = Teacher("Ms. Sharma", 42, "T9876", "Mathematics")


print("\n--- Teacher Actions ---")
teacher1.introduce()
teacher1.teach()


print(f"\nDirect access to protected attribute (discouraged): {teacher1._subject}")
```
