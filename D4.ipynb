```python
# Define the dictionary
intro = {
    "name": "",
    "age": "",
    "education": ""
}

# --- 1. Loop over keys ---
print("Keys:")
for key in intro:
    print(key)

# --- 2. Loop over values ---
print("\nValues:")
for value in intro.values():
    print(value)

# --- 3. Loop over key-value pairs ---
print("\nKey-Value Pairs:")
for key, value in intro.items():
    print(f"{key}: {value}")

# --- 4. Optional: Fill dictionary using user input ---
print("\nFill the dictionary:")
for key in intro:
    intro[key] = input(f"Enter your {key}: ")

print("\nUpdated Dictionary:")
for key, value in intro.items():
    print(f"{key}: {value}")
```

```python
import random
secret_number =random.randint(1, 100)
```

```python
attempt =0
guess=0

print("Guess the number between 1 and 100!")

for attempt in range(1, 100):  
    print(f"Attempt 🌀{attempt}")
    guess = int(input("Your guess: "))
    
    if guess == secret_number:
        print(f"⚠️Congrats! You guessed it in {attempt} attempts.")
        break
    elif guess < secret_number:
        print("Too low, try again.")
    else:
        print("Too high, try again.")
```
