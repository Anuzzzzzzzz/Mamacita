```python
import requests
from bs4 import BeautifulSoup
import csv
```

```python
url = "https://www.thebritishcollege.edu.np"
response = requests.get(url)
soup = BeautifulSoup(response.content, "html.parser")
```

```python
print("Website Title:")
print(soup.title.text.strip())
```

```python
print("Headings on the Page:")
headings = soup.find_all("h2")
for h in headings:
    print("-", h.text.strip())
```

```python
print("Paragraphs on the Page:")
paragraphs = soup.find_all("p")
for p in paragraphs:
    print( p.text.strip() )
```

```python
courses = soup.find_all("div", {"class": "course"})
for c in courses:
    print(c.text.strip())
```

```python
courses = soup.select("li > a")  # all links inside <li>
course_list = [c.text.strip() for c in courses if c.text.strip() != ""]
print("Courses / Menu Items:")
for c in course_list:
    print("-", c)
```

```python
academic_courses = [c for c in course_list if any(x in c for x in ["BBA", "MBA", "BSc", "MSc"])]
print("Academic Courses:")
for ac in academic_courses:
    print("-", ac)
```

```python
with open("tbc_courses.csv", "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["Course Name"])
    for ac in academic_courses:
        writer.writerow([ac])
print("✅ Saved academic courses to tbc_courses.csv")
```
