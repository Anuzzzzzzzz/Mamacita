```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from bs4 import BeautifulSoup
from urllib.parse import urljoin
import time

# Chrome browser options
chrome_options = Options()
chrome_options.add_argument("--headless=new")  # background ma run garna ko lagi

driver = webdriver.Chrome(options=chrome_options)  # start browser

base_url = "https://news.ycombinator.com/"
current_page_url = base_url

all_news_data = []  # store all news titles
page_count = 0
max_pages = 5  # max pages to scrape

try:
    while current_page_url and page_count < max_pages:
        page_count += 1
        print(f"\nPage {page_count}:\n")  # show current page

        driver.get(current_page_url)

        # wait until news items are loaded
        WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.CLASS_NAME, "athing"))
        )

        soup = BeautifulSoup(driver.page_source, "lxml")

        # get all news items on page
        rows = driver.find_elements(By.CLASS_NAME, "athing")
        for row in rows:
            title = row.find_element(By.CLASS_NAME, "titleline").text
            print("Title:", title)  # print title
            all_news_data.append(title)

        # check if there is a "More" link for next page
        more_link = soup.find("a", class_="morelink")
        if more_link:
            current_page_url = urljoin(base_url, more_link["href"])
            time.sleep(1)  # small wait before next page
        else:
            current_page_url = None  # no more pages

finally:
    driver.quit()
    print("\nBrowser closed.")

# optional: print all news collected
print("\nAll Hacker News Titles:\n")
for title in all_news_data:
    print(title)
```
