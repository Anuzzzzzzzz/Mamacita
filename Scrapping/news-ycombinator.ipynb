```python
from selenium import webdriver
from selenium.webdriver.chrome.options import Options
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from bs4 import BeautifulSoup
from urllib.parse import urljoin
import time

chrome_options = Options()
chrome_options.add_argument("--headless=new")



driver = webdriver.Chrome(options=chrome_options)

base_url = "https://news.ycombinator.com/"
current_page_url = base_url

all_news_data = []
page_count = 0
max_pages = 5

try:
    while current_page_url and page_count < max_pages:
        print("Page:", page_count + 1)
        page_count += 1

        driver.get(current_page_url)

        WebDriverWait(driver, 10).until(
            EC.presence_of_element_located((By.CLASS_NAME, "athing"))
        )

        soup = BeautifulSoup(driver.page_source, "lxml")

        rows = driver.find_elements(By.CLASS_NAME, "athing")
        for row in rows:
            title = row.find_element(By.CLASS_NAME, "titleline").text
            all_news_data.append(title)

        more_link = soup.find("a", class_="morelink")
        if more_link:
            current_page_url = urljoin(base_url, more_link["href"])
            time.sleep(1)
        else:
            current_page_url = None

finally:
    driver.quit()
    print("\nBrowser closed.")

print("\nHacker News Titles:\n")
for title in all_news_data:
    print(title)
```
