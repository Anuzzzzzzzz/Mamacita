```python
# Task 1: The messy_data Cleaner
# Focus: String Methods, Slicing, F-Strings
# Scenario: You have received raw text data from a user input, but it is messy. It has extra spaces, inconsistent capitalization, and contains a forbidden word.
# Create a variable raw_text with this exact value:
# " PyThOn is AweSome and badword programming! "
# Clean the text:
# Remove the leading and trailing spaces.
# Convert the entire string to lowercase.
# Replace the word "badword" with "powerful".
# Slice it: Extract the word "programming" from your cleaned text using slicing syntax [start:stop].
# Report: Use an f-string to print the final result in this format:
# "Original length: {x}, Cleaned text: '{y}', Extracted word: '{z}'"

raw_text = " PyThOn is AweSome and badword programming! "
cleaned_text = raw_text.strip().lower().replace("badword", "powerful")
extracted_word = cleaned_text[31:42]
print(f"Original length: {len(raw_text)}, Cleaned text: '{cleaned_text}', Extracted word: '{extracted_word}'")
```

```python
# Task 2: The Dynamic Shopping Cart
# Focus: Lists (Mutable, Ordered), Loops
# Scenario: You are building the backend for a shopping cart application.
# Create an empty list called cart.
# Add items: Use a loop to append "Apple", "Banana", and "Milk" to the cart.
# User Change: The user decides they don't want "Banana". Write code to check if "Banana" is in the list, and if so, remove it.
# Insert Item: The user forgot "Bread". Insert "Bread" at index 1 (so it sits between Apple and Milk).
# Sort & Print: Sort the list alphabetically and print the final items using a for loop, printing each item on a new line.

# 1. Create an empty list
cart = []

# 2. Add items using a loop (Fixed the missing quote after "Apple")
items_to_add = ["Apple", "Banana", "Milk"]
for item in items_to_add:
    cart.append(item)  # Indented this line

# 3. User Change: Check and remove "Banana"
if "Banana" in cart:
    cart.remove("Banana")  # Indented this line

# 4. Insert Item: "Bread" at index 1 (This was missing in your snippet)
cart.insert(1, "Bread")

# 5. Sort the list alphabetically
cart.sort()

# 6. Print final items using a loop
for item in cart:
    print(item)  # Indented this line
```

```python
# Task 3: GPS Coordinate Tracker
# Focus: Tuples (Immutable, Unpacking)
# Scenario: You are working on a mapping application. Coordinates should never be modified once recorded, so you must use Tuples.
# Create a list of tuples representing locations:
# locations = [("Home", (27.71, 85.32)), ("Office", (27.72, 85.33))]
# Loop through the list. Inside the loop, use tuple unpacking to separate the data into variables: name, lat, and lon.
# Print a formatted string for each location:
# "Location: Home | Latitude: 27.71 | Longitude: 85.32"

locations = [
    ("Home", (27.71, 85.32)),
    ("Office", (27.72, 85.33))
]

for name, (latitude, longitude) in locations:
    print(f"Location: {name} | Latitude: {latitude} | longitude: {longitude}")
```

```python
# Task 4: The Email De-Duplicator
# Focus: Sets (Unique, Unordered)
# Scenario: You have two mailing lists for a newsletter. Some users are on both lists, and you want to send an email only once per person.
# Create two lists with overlapping names:
# list_a = ["alice@example.com", "bob@example.com", "charlie@example.com"]
# list_b = ["bob@example.com", "david@example.com", "alice@example.com","boby@example.com", "david@example.c", "alice@example.cm"]
# Convert both lists to Sets.
# Union: Create a new set called unique_emails that contains all users from both lists (duplicates removed automatically).
# Intersection: Create a set called loyal_users containing only emails found in both lists.
# Print the count of unique emails using len().

list_a = ["alice@example.com", "bob@example.com", "charlie@example.com"]
list_b = ["bob@example.com", "david@example.com", "alice@example.com", 
          "boby@example.com", "david@example.c", "alice@example.cm"]

#converting list to set
set_a = set(list_a)
set_b = set(list_b)

#Union(cobining bith sets)
unique_emails = set_a.union(set_b)

#intersection(find email that exist in both sets)
loyal_users =set_a.intersection(set_b)

print(f"total unique emails: {len(unique_emails)}")
print(f"Unique emails: {unique_emails}")
print(f"Loyal users (in both lists): {loyal_users}")
```

```python
# ask 5: The "All-in-One" Inventory System
# Focus: Capstone (Combining Lists, Sets, Dicts, Loops, String Methods)
# Scenario: You manage a grocery store inventory. The input data is messy (duplicates, mixed case), and you need to structure it.
# You are given this raw list:
# raw_products = [" apPle ", "BANANA", " apple ", "orange", "Banana", " cherry ", "APricot", "guava ", "orange", "Banana"]
# Clean & Unique:
# Create an empty set.
# Loop through raw_products, strip whitespace, and convert them to lowercase. Add the clean strings to your set.
# Inventory Count:
# Create an empty dictionary inventory.
# Loop through your clean unique set. Assign a stock quantity of 50 to every item.
# Store it in the dictionary (e.g., {'apple': 50}).
# Report:
# Iterate through the dictionary and print the inventory in this format:
# "Product: Apple | Stock: 50"
# Hint: Use .title() on the product name when printing to make it look nice.

# Raw input data
raw_products = [" apPle ", "BANANA", " apple ", "orange", "Banana", " cherry ", "APricot", "guava ", "orange", "Banana"]

# 1. Clean the names
clean_products = set()
for item in raw_products:
    clean_products.add(item.strip().lower()) 

# 2. Create the inventory
inventory = {} 
for product in clean_products:
    inventory[product] = 50

# 3. Print the report in ALPHABETICAL order
# We add sorted() here to arrange the keys from A to Z
print("--- Inventory Report ---")
for product in sorted(inventory):
    print(f"Product: {product.title()} | Stock: {inventory[product]}")
    
# 4. Inventory Count (Total unique items)
print("-" * 24)
print(f"Total Unique Products: {len(inventory)}")
    
```
